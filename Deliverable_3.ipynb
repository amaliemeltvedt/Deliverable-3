## Character class
The Character class represents a character in the RPG. It contains the attributes name, max HP, damage and weapon. The name is a string chosen by the player themselves, while the HP max attribute is an integer with a default value of 100. This means the character's hit points cannot surpass the given max. Damage is an integer with a default value of 10, meaning the character deals 10 damage if it does not have a weapon equipped. The last attribute is the weapon attribute, which represents a chosen weapon or item from the Weapon class. It will either deal a random amount of damage decided by the dice throw, or the character can choose not to have a weapon and deal the default damage value.
