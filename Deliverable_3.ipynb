## Character class
The `Character` class represents a character in the RPG. It contains the attributes name, max HP, damage and weapon. The name is a string chosen by the player themselves, while the HP max attribute is an integer with a default value of 100. This means the character's hit points cannot surpass the given max. Damage is an integer with a default value of 10, meaning the character deals 10 damage if it does not have a weapon equipped. 

For example, a player can create their character by only assigning a value to the `name` attribute. This way, the `Character` object will be automatically assigned the default HP and damage values.

```python
# Creating a character
hero = Character("Legolas")

# Displaying the character
print(hero)
hero.introduce()
```

As seen in the output, our created Character object has the default HP value of 100 and 10 damage. However, the created character can have any value of HP and damage if the player chooses to overwrite the default attribute values.

```python
# Creating a character with custom HP and damage
enemy = Character("Sauron", hp_max = 150, damage = 20)

print(enemy)
```

The `Character` class contains methods for taking and dealing damage. The `take_damage()` method ensures only the character's own HP is affected when attacked, while the `attack()` method calls to the opponent's `take_damage()` method. For example, our Hero can deal damage on their enemy by calling the `attack()` method.

```python
# Hero attacks enemy
hero.attack(enemy)

print("After attack:")
print(enemy)

# Enemy attacks hero
enemy.attack(hero)

print("After attack:")
print(hero)
```

The `take_damage()` method could also be used to reduce any amount of HP. To ensure that the character's HP cannot go below zero, we added a lower boundary of 0 HP using the `max()` function.

```python
# Reducing a custom amount of HP
hero.take_damage(30)

print("After taking damage:")
print(hero)

# Trying to reduce to a negative number will still result in 0 HP
hero.take_damage(100)

print("After taking too much damage:")
print(hero)
```

A character can also heal itself using the `heal()` method.

```python
# Character heals
hero.heal(20)

print("After healing:")
print(hero)
```

The `Character` class contains an `is_alive` property to check if the Character is still alive. This will return True if the Character's HP is above 0, and false if it is 0.

```python
print(hero.is_alive)
```

## Dice class

The Dice class represents a dice with a specified number of sides. The player can choose the number of sides when creating the dice object. 

The class has one attribute, `sides`, which is an interger with a default value of 6. 

Dice class has three methods:
-  `roll()`: Rolls the dice and returns a random number between 1 and the number of sides on the dice. 
- `__repr__()`: Defines how the dice object is represented as a string. 
- `__call__()`: Makes it possible to use the Dice object as a function by calling `self.roll()`.

We can create a Dice object without defining the number of sides, to controll that the default sides will be 6. We can also define the number of sides to 20 as an example:

```python
# Creating dice objects
dice = Dice()
dice_20 = Dice(20)

print(dice)
print(dice_20)
```

We have ensured that the dice at least must have 2 sides. If the player defines a dice with 1 side, it will give a ValueError.

```python
# Try to create a dice with only one side
try:
    invalid_dice = Dice(sides=1)
except ValueError as error:
    print(error)
```

To test the `roll()` function we can use an example dice with 20 sides, and controll that the function can give different results for each roll:

```python
# Rolling dice_20 five times
for i in range(5):
    print(dice_20.roll())
```

The method `__repr__()` decides how the object will be represented with string, when we use `repr()`. Again we can use dice_20 as an example:

```python
# Printing the representation of dice_20
print(repr(dice_20))
```

We have implemented a `__call__()` method in the Dice class. This makes it possible to only write the name of the dice object, for example `dice()`, rather than `dice.roll()`. Both `dice.roll()` and `dice()` will roll the dice.

```python
# Rolling the dice using both the roll method and the __call__ method

result_using_roll = dice_20.roll()
print(f"Result using roll: {result_using_roll}")

result_using_call = dice_20()
print(f"Result using __call__: {result_using_call}")
```

## Weapon class

The Weapon class represents weapons that characters can be equipped with during a battle. These weapons can deal damage to other characters. The class has two attributes, `name` and `damage_dice`. `Name` stores the name of the weapon given by the player, while `damage_dice` stores a Dice object that determines the weapon's damage.

This class uses the method `roll_damage()` which rolls the weapon's damage dice, and returns a damage value.

Composition is used in the Weapon class because a Weapon object contains a Dice object. The Weapon class use the `roll()` method of the Dice class to determine the weapon's damage.

We can create a Dice object and pass it as the damage dice when creating a Weapon object:

```python
# Creating a Dice object with 10 sides
spear_dice = Dice(sides=10)

# Creating a Weapon object containing the Dice object (composition)
spear = Weapon(name="Spear", damage_dice=spear_dice)

print(spear.name)
print(spear.damage_dice)
```

We can then use the `roll_damage()` method on the Weapon object, in this example `spear`, to determine how much damage the weapon deals:

```python
# Rolling the damage dice
damage_by_spear = spear.roll_damage()

print(f"The {spear.name} deals {damage_by_spear} damage.")
```

This can be implied on multiple weapons, for example:

```python
# Create different dices
axe_dice = Dice(sides=20)
bow_dice = Dice(sides=12)

# Create weapons using the dices
axe = Weapon("Axe", axe_dice)
bow = Weapon("Bow", bow_dice)

# Roll damage for each weapon
print(f"Axe damage: {axe.roll_damage()}")
print(f"Bow damage: {bow.roll_damage()}")
```

## Character subclasses

The `Character` class contains to subclasses of specialised Characters:

- `Warrior` subclass
- `Mage` subclass

### Warrior subclass
The `Warrior` subclass is a defensive specialist subclass, equipped with armor to reduce the incoming damage before taking it. It inherits the attributes from the regular `Character` class, then adds an `armor` attribute. This attribute is an integer with a default value of 5. This means an incoming attack performed by the enemy will be reduced by 5 before it hits the Warrior.

```python
#Creating a Warrior
warrior = Warrior("Gimli")

print(warrior)
print(f"{warrior.name} has an armor of {warrior.armor}")
```

If Gimli gets attacked, his armor will reduce the incoming damage by 5. 

```python
# Gimli takes less damage 
print(f"HP before attack: {warrior.hp}")
print(f"Armor reduction: {warrior.armor}")

warrior.take_damage(20)

print(f"Damage taken: {20-5}")
print(f"HP after attack: {warrior.hp}")
```

As we can see, Gimli's armor reduced the damage from 20 to 15.

### Mage subclass
The Mage subclass is a resource-based attacker which inherits the standard `Character` attributes, but gains two more: `mana` and `mana_max`.


```python
# Creating the Mage
mage = Mage("Gandalf")

print(mage)
print(f"{mage.name} has {mage.mana}/{mage.mana_max} mana")
```

Mana is an energy source allowing the Mage to deal a magical spell attack dealing double damage. Such an attack costs 10 mana and does not use any weapon, even if the Mage has one.

```python
# Showcasing the magical attack
print(f"{enemy.name} HP before attack: {enemy.hp}")
print(f"{mage.name} mana before attack: {mage.mana}")

mage.attack(enemy)

print(f"{enemy.name} HP after attack: {enemy.hp}")
print(f"{mage.name} mana after attack: {mage.mana}")
```

If we manually change the Mage's amount of mana to 5, meaning it is below the required amount for a magical attack, it will fall back to the inherited base `attack()` method. The amount of HP the attack will reduce is determined by the weapon (if equipped) and dice throw, or the default damage. This will in turn restore 5 mana.

```python
# Changing the mage's mana
mage.mana = 5

print(f"{enemy.name} HP before attack: {enemy.hp}")
print(f"{mage.name} before attack: {mage.mana}")


mage.attack(enemy)

print(f"{enemy.name} HP after attack: {enemy.hp}")
print(f"{mage.name} after attack: {mage.mana}")
```

## Battle class
Lastly, we created a Battle class to manage turn-based combat between two `Character` objects. The class has two attributes, `hero` and `enemy`, which both store a `Character` object. In addition to the initial method, it also contains the two methods `play_round()` and `simulate_battle()`. The `play_round()` method describes the actions during a single round of combat, where the `hero` first attacks the `enemy`, and if `enemy` is still alive, then `enemy` will in turn attack `hero`. The `simulate_battle()` method contains a `while` loop that repeats the `play_round()` actions until one character's `is_alive` property returns False.

For example, hero and enemy characters can be made into a `battle` object.

```python
# Creating a new final hero with max HP
hero = Character("Frodo")

# Overwriting the enemy variable to make sure Sauron also has max HP
enemy = Character("Sauron", hp_max = 150)

battle = Battle(hero, enemy)

print(battle.hero)
print(battle.enemy)
```

In this example, Frodo functions as the hero and Sauron as the enemy. By calling the `play_round()` method, exactly one round of combat will be executed.

```python
print(f"Before round: {battle.hero.name} HP: {battle.hero.hp}, {battle.enemy.name} HP: {battle.enemy.hp}")

battle.play_round()
```

To simulate an entire battle, we can call the `simulate.battle()` method.

```python
battle.simulate_battle()
```

As we see, the battle is finished and Sauron unfortunately wins. Assigning a subclass and a weapon to the characters might change the outcome. We can overwrite the `Character` variables, this time giving them a weapon and a subclass.

```python
# Creating a dice for the weapons sword and staff
sword_dice = Dice(sides=10)
staff_dice = Dice(sides=12)

# Create Weapon objects
sword = Weapon(name="Sword", damage_dice=sword_dice)
staff = Weapon(name="Staff", damage_dice=staff_dice)

# Create Frodo as a Warrior with a sword
# This creates a new Warrior object, but we give the same hp_max value as before
hero = Warrior(
    name="Frodo",
    hp_max=100,
    armor=10,
    weapon=sword
)

# Create Sauron as a Mage with a magic staff
# This creates a new Mage object, but we give the same hp_max value as before
enemy = Mage(
    name="Sauron",
    hp_max=110,
    mana=50,
    weapon=staff
)

# Display the character information
print("Battle participants:")
print(f"{hero.name}: Warrior with {hero.hp} HP, "
      f"{hero.armor} armor and a {hero.weapon.name}"
      )

print(f"{enemy.name}: Mage with {enemy.hp} HP, "
      f"{enemy.mana} mana and a {enemy.weapon.name}")
```

Now that we have created characters with subclasses and weapons, we can again use `play_round()` to simulate one round of combat.

```python
battle = Battle(hero, enemy)

# Display status before one round
print("\nBefore round:")
print(f"Frodo HP: {hero.hp}, Armor: {hero.armor}")
print(f"Sauron HP: {enemy.hp}, Mana: {enemy.mana}")

# Play one round
battle.play_round()
```

After one round we see that Frodo lost exactly 10 HP, while Sauron's mana has decreased with 10. This means he used his mana energy to double the base damage (5) dealt on Frodo. The damage Frodo dealt on Sauron will vary depending on the die throw because he has a weapon equipped. In a full battle, Sauron's mana will run out and his base attack will also vary depending on a die throw.

We can simulate the entire battle to find a winner using the `simulate_battle()` method:

```python
# Simulate the entire battle
battle.simulate_battle()
```

As the damage dealt by the weapon equipped is decided by a random die throw, the results will vary.
