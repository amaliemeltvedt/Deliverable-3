## Character class
The Character class represents a character in the RPG. It contains the attributes name, max HP, damage and weapon. The name is a string chosen by the player themselves, while the HP max attribute is an integer with a default value of 100. This means the character's hit points cannot surpass the given max. Damage is an integer with a default value of 10, meaning the character deals 10 damage if it does not have a weapon equipped. 

For example, a player can create their character by only assigning a value to the `name` attribute. This way, the Character object will be automatically assigned the default HP and damage values.

```python
# Creating a character
hero = Character("Legolas")

# Displaying the character
print(hero)
hero.introduce()
```

As seen in the output, our created Character object has the default HP value of 100 and 10 damage. However, the created character can have any value of HP and damage if the player chooses to overwrite the default attribute values.

```python
# Creating a character with custom HP and damage
enemy = Character("Sauron", hp_max = 150, damage = 20)

print(enemy)
```

The `Character` class contains methods for taking and dealing damage. The `take_damage()` method ensures only the character's own HP is affected when attacked, while the `attack()` method calls to the opponent's `take_damage()` method. For example, our Hero can deal damage on their enemy by calling the `attack()` method.

```python
# Hero attacks enemy
hero.attack(enemy)

print("The enemy now has:")
print(enemy)

# Enemy attacks hero
enemy.attack(hero)

print("The hero now has:")
print(hero)
```

A character can also heal itself using the `heal()` method.

```python
# Character heals
hero.heal(20)

print("The character now has:")
print(hero)
```
